```python
import pandas as pd
import pickle
import torch 
import joblib
```

```python
des_em_array = joblib.load("description_embedding.pkl")
description_embedding = torch.tensor(des_em_array)
```

```python
rev_em_array = joblib.load("reviews_embedding.pkl")
reviews_embedding = torch.tensor(rev_em_array)
```

```python
des_df = pd.read_csv('book_descriptions.csv')
reviews_df = pd.read_csv('processed_reviews.csv')
```

```python
import random

# Pick a random index for reviews 

random_idx = random.randint(0, len(reviews_embedding) - 1)
query_embedding = reviews_embedding[random_idx] # Use this as a query to find most similar descriptions 
random_review = reviews_df.iloc[random_idx]['cleaned_reviews'] 

print(f"Random review query: {random_review}\n")
```

```python
import torch
import torch.nn.functional as f
```

```python
# Normalize

#des_em_normalized = f.normalize(description_embedding, dim=1)
#query_em_normalized = f.normalize(query_embedding, dim=0) 
```

```python
# Compute cosine similarity

cos_sim = torch.matmul(description_embedding, query_embedding)
```

```python
# Get top-k similar descriptions
k = 5
top_k = torch.topk(cos_sim, k=k)

# Results
for i, (score, idx) in enumerate(zip(top_k.values, top_k.indices)):
    idx = idx.item()
    book_title = des_df.iloc[idx]['title']
    book_description = des_df.iloc[idx]['cleaned_description']
    book_cover = des_df.iloc[idx]['image_url']

    print(f"Rank {i+1}  Cos_Sim Score: {score:.4f}")
    print(f"Book Title: {book_title}")
    print(f"Book description: {book_description}")
    print(f"Book cover: {book_cover}")
    print("-------------------------------")
```
